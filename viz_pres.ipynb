```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
authors = {'authors' : 68,
           'male' : 66,
           'female' : 4,
           'unknown' : 2}
```

```python
# Data to plot
labels = ['Male', 'Female', 'Unknown']
sizes = [authors['male'], authors['female'], authors['unknown']]
colors = ['lightblue', 'lightcoral', 'lightgray']
explode = (0.1, 0, 0)  # explode the first slice (i.e., 'Male')

# Plotting the pie chart
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
        shadow=True, startangle=140)

# Equal aspect ratio ensures that pie is drawn as a circle.
plt.axis('equal')

# Title of the pie chart
plt.title('Distribution of Authors in the SrpELTeC Collection by Gender')

# Display the plot
plt.show()
```

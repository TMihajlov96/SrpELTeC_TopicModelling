```python
data_path = '/Users/teamihajlov/FEDCIS2024/CLIB_TopicVisualization/data_clean_final.csv'
data = pd.read_csv(data_path)
data.drop(columns = ["Unnamed: 0"], inplace = True)
data.head()
```

```python
corpus_data = data['texts_clean'].tolist()
print(corpus_data[0])
print(len(corpus_data))
```

```python
# Converting clean texts to a list
corpus_data = data['texts_clean'].tolist()
print(corpus_data[0])
print(len(corpus_data))
```

```python
# Setting up NMF parameters
vectorizer = CountVectorizer(min_df = 15, max_df = 0.8, stop_words=stopwords)
topic_model = NMF(n_components = 7)

pipe_NMF = Pipeline(steps=[
    ('vectorizer', vectorizer),
    ('model',  topic_model),])
```

```python
data_topics = pipe_NMF.fit_transform(corpus_data)
print(data_topics[0])
```

```python
dictionary_texts = vectorizer.get_feature_names_out()
print(dictionary_texts[:100]) # displaying first 100 terms
print(len(dictionary_texts))
```

```python
def top_words(svd, vocab):
    for index, component in enumerate(svd.components_):
        zipped = zip(vocab, component)
        top_terms_key=sorted(zipped, key = lambda t: t[1], reverse = True)[:10]
        top_terms_list=list(dict(top_terms_key).keys())
        final = print("Topic "+str(index)+": ",top_terms_list)
    return(final)
```

```python
top_words(topic_model, dictionary_texts)
```

```python
topic_words = topic_model.components_
topic_words
```

```python
import gensim
from gensim.models.coherencemodel import CoherenceModel
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Ensure you have the NLTK data files
nltk.download('punkt')
nltk.download('stopwords')

# Extract top words for each topic
num_top_words = 10
topic_top_words = []
for topic_idx, topic in enumerate(topic_words):
    top_words = [dictionary_texts[i] for i in topic.argsort()[:-num_top_words - 1:-1]]
    topic_top_words.append(top_words)

# Prepare documents for coherence calculation
stop_words = set(stopwords.words('english'))
texts = [[word for word in word_tokenize(doc.lower()) if word.isalpha() and word not in stop_words] for doc in corpus_data]

# Create a dictionary and corpus for Gensim
dictionary = gensim.corpora.Dictionary(texts)
corpus = [dictionary.doc2bow(text) for text in texts]

# Calculate Coherence Score
coherence_model = CoherenceModel(topics=topic_top_words, texts=texts, dictionary=dictionary, coherence='c_v')
coherence_score = coherence_model.get_coherence()
print(f'Coherence Score: {coherence_score}')
```

```python
# Calculate Topic Diversity
num_topics = 7
unique_words = set()
for topic in topic_top_words:
    unique_words.update(topic)

# Total unique words across all topics
total_unique_words = len(unique_words)

# Average number of unique words per topic
avg_unique_words_per_topic = total_unique_words / num_topics

# Topic Diversity Score
topic_diversity = avg_unique_words_per_topic / num_top_words
print(f'Topic Diversity: {topic_diversity}')
```
